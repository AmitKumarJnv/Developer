```python
# is, is not, ==, != operators
list_a = [1, 2, 3]
list_b = [1, 2, 3]
list_c = [4, 5, 6]

# Comparing values
print(list_a == list_b)  # True, because the contents are identical
print(list_a != list_c)  # True, because [1, 2, 3] is not [4, 5, 6]

list_a = [1, 2, 3]
list_b = [1, 2, 3]
list_d = list_a  # We are telling list_d to point to the exact same object as list_a

# Comparing identity (memory addresses)
print(list_a is list_b)      # False. They have the same values, but are two distinct objects.
print(list_a is not list_b)  # True. They live in different places in memory.
print(list_a is list_d)      # True. list_d is just another name for list_a.

print(id(list_a))  # e.g., 140321586734080
print(id(list_b))  # e.g., 140321586734144 (Different address)
print(id(list_d))  # e.g., 140321586734080 (Same as list_a)
```

### Iterators, Generators, and Decorators

```python
my_list = [10, 20, 30]

# Get an iterator object from the list
iterator_obj = iter(my_list)  # Calls my_list.__iter__()

# Manually step through it
print(next(iterator_obj))  # Output: 10 (calls __next__)
print(next(iterator_obj))  # Output: 20
print(next(iterator_obj))  # Output: 30

# The next call raises StopIteration, signaling the end
try:
    print(next(iterator_obj))
except StopIteration:
    print("The stream is exhausted!")
```

```python
class Countdown:
    def __init__(self, start):
        self.current = start

    def __iter__(self):
        return self  # The object itself is its own iterator

    def __next__(self):
        if self.current <= 0:
            raise StopIteration  # End of sequence
        value = self.current
        self.current -= 1
        return value

# Usage
countdown = Countdown(5)
for num in countdown:
    print(num)  # Output: 5, 4, 3, 2, 1

# WARNING: This iterator is now exhausted!
print(list(countdown))  # Output: [] (empty, because it's already at 0)
```

```python
class Getting_Even_Numbers:
    def __init__(self,number):
        self.start_point = 0
        self.even_number_count = number

    def __iter__(self):
        return self    

    def __next__(self):
        if self.even_number_count <=0:
            raise StopIteration
        value = self.start_point
        self.even_number_count -=1
        self.start_point +=2
        return value
```

```python
def fibonacci_series_numbers(number:int):
    index, first_num, second_num = 0,0,1

    while index < number:
        yield first_num
        first_num,second_num = second_num, first_num + second_num
        index += 1
```

```python
fib_num_4 = fibonacci_series_numbers(4)
```

```python
def my_decorator(func):
    def wrapper(*args, **kwargs):
        print(f"Before:{func.__name__},args:{args},kwargs:{kwargs}")
        result = func(*args, **kwargs)
        print("After")
        return result
    return wrapper

@my_decorator
def add(a, b):
    return a + b

print(add(3, 5))
```

```python
def my_multiplication_decorator(func):
    def wrapper(*args,**kwargs):
        print(f"function name: {func.__name__},function arguments: {args,kwargs}, START")
        value = func(*args,**kwargs)
        print(f"function name: {func.__name__},function arguments: {args,kwargs}, END")
        return value
    return wrapper  

@my_multiplication_decorator
def multi(a,b):
    return a*b

multi(3,8)
```

```python
import functools

def my_decorator(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        return func(*args, **kwargs)
    return wrapper

@my_decorator
def greet(name):
    """Greet someone by name."""
    return f"Hello, {name}"

print(greet.__name__)   # greet
print(greet.__doc__)    # Greet someone by name.
```

```python
def uppercase(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        result = func(*args, **kwargs)
        return result.upper() if isinstance(result, str) else result
    return wrapper

def add_greeting(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        result = func(*args, **kwargs)
        return f"Hello, {result}" if isinstance(result, str) else result
    return wrapper

@uppercase
@add_greeting
def get_name():
    return "alice"

print(get_name())  # HELLO, ALICE
```

```python
def repeat(times):
    def decorator(func):
        @functools.wraps(func)
        def wrapper(*args, **kwargs):
            for _ in range(times):
                result = func(*args, **kwargs)
            return result
        return wrapper
    return decorator

@repeat(3)
def say_hi():
    print("Hi")

say_hi()
```

```python
import time
import functools

def timer(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start = time.perf_counter()
        result = func(*args, **kwargs)
        end = time.perf_counter()
        print(f"{func.__name__} took {end - start:.6f} seconds")
        return result
    return wrapper

@timer
def slow_function():
    time.sleep(0.5)

slow_function()
```

```python
import functools
import logging

logging.basicConfig(level=logging.INFO)

def log_call(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        logging.info(f"Calling {func.__name__} with args={args}, kwargs={kwargs}")
        result = func(*args, **kwargs)
        logging.info(f"{func.__name__} returned {result}")
        return result
    return wrapper

@log_call
def multiply(x, y):
    return x * y

multiply(3, 4)
```

```python
import time
from datetime import datetime

# --- USING perf_counter (CORRECT for benchmarking) ---
start = time.perf_counter()
time.sleep(1)  # Simulate work
end = time.perf_counter()

duration = end - start
print(f"Work took {duration:.9f} seconds")  # Output: ~1.000000123 (accurate)

# --- USING time.time() (WRONG for benchmarking) ---
start_wall = time.time()
time.sleep(1)
end_wall = time.time()
print(f"Wall time difference: {end_wall - start_wall}")  # Usually fine, but can be negative if an NTP sync occurs mid-measurement!

# --- USING datetime.now() (for logging) ---
now = datetime.now()
print(f"Current timestamp for log: {now}")  # e.g., 2026-08-11 10:30:45.123456
```

```python
# Standard positional arguments
# *args
# Standard keyword arguments (with default values)
# **kwargs

def create_report(title, *args, author="Anonymous", **kwargs):
    print(f"Title: {title}")
    print(f"Author: {author}")
    
    print("\nChapters:")
    for chapter in args:
        print(f"- {chapter}")
        
    print("\nMetadata:")
    for key, value in kwargs.items():
        print(f"{key}: {value}")

# Calling the function with a mix of everything
create_report(
    "Annual Sales",          # Standard argument (title)
    "Q1", "Q2", "Q3",        # *args (packed into a tuple)
    author="Jane Doe",       # Overriding the default keyword argument
    confidential=True,       # **kwargs (packed into a dict)
    pages=42                 # **kwargs (packed into a dict)
)
```

```python
import logging
import time
from functools import wraps

logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

def log_and_time(func):
    """A production decorator to track execution time and log calls."""
    @wraps(func)
    def wrapper(*args, **kwargs):
        start_time = time.perf_counter()
        
        try:
            # We don't know what arguments 'func' needs, 
            # so we just unpack and forward everything we received.
            result = func(*args, **kwargs)
            return result
        finally:
            execution_time = time.perf_counter() - start_time
            logger.info(f"[{func.__name__}] executed in {execution_time:.4f}s")
            
    return wrapper

# Now we can safely apply this to functions with totally different signatures
@log_and_time
def process_payment(user_id, amount, currency="USD"):
    time.sleep(0.5) # Simulating a network call
    return True

@log_and_time
def fetch_user_data():
    time.sleep(0.2)
    return {"id": 1, "name": "Alice"}

process_payment(101, 50.00, currency="EUR")
fetch_user_data()
```
